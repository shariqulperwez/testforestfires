```python
# Ordinal Encoding for the column -- Total Stops

df.head(2)
```

```python
df['Total_Stops'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Instance creation
oe = OrdinalEncoder(categories=[['non-stop','1 stop','2 stops','3 stops','4 stops']])

# fit and transform
oe_enc = oe.fit_transform(df[['Total_Stops']])
oe_enc
```

```python
# converting into a dataframe

df_encoded = pd.DataFrame(data=oe_enc, columns=['Stops'])
df_encoded
```

```python
# concatination

df = pd.concat([df,df_encoded], axis=1)
```

```python
# One hot Encoding for Columns Airline, Source, Destination and Additional_Info

from sklearn.preprocessing import OneHotEncoder

#instance
ohe=OneHotEncoder()

#fit and transform
enc = ohe.fit_transform(df[['Airline', 'Source', 'Destination', 'Additional_Info']]).toarray()
```

```python
# Converting to a dataframe

df_ohe = pd.DataFrame(data=enc, columns=ohe.get_feature_names_out())
df_ohe
```

```python
# concatination

df = pd.concat([df, df_ohe], axis=1)
```
