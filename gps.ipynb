3) How many apps are there on Google Play store which get 5 rating?

```python
data = df.groupby('Category')['Rating'].value_counts().reset_index()

data = data.rename(columns={'Category': 'Category','Rating': 'Rating','count':'Count'})
```

```python
rating = [5]
for rate in rating:
    df_rated = data[data['Rating']==rate]
    print(df_rated[['Category','Count']])
```

```python
rating = df.groupby(['Category','Installs', 'App'])['Rating'].sum().sort_values(ascending = False).reset_index()
rating = rating.sort_values('Installs', ascending=False)
toprating_apps = rating[rating.Rating == 5.0]
toprating_apps
# top_rate = rating[rating['Rating']==5.0]
# top_rate.head(1)
print("Number of 5 rated apps",toprating_apps.shape[0])
toprating_apps.head(1)

# toprating_apps
# top_rate
```
